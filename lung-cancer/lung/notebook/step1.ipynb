```python
import numpy as np
from pyteomics import mzml

MZ_MIN = 300
MZ_MAX = 1800
DELTA_M = 1.0

N_BINS = int((MZ_MAX - MZ_MIN) / DELTA_M)
print("Total bins:", N_BINS)
```

```python
def bin_file(path):
    vec = np.zeros(N_BINS, dtype=np.float64)

    with mzml.MzML(path) as reader:
        for spec in reader:
            if spec.get("ms level") != 1:
                continue

            mzs = spec.get("m/z array")
            ints = spec.get("intensity array")

            if mzs is None or len(mzs) == 0:
                continue

            mask = (mzs >= MZ_MIN) & (mzs < MZ_MAX)
            mzs = mzs[mask]
            ints = ints[mask]

            idx = ((mzs - MZ_MIN) / DELTA_M).astype(int)
            np.add.at(vec, idx, ints)

    # Normalize (TIC)
    if vec.sum() > 0:
        vec = vec / vec.sum()

    # Log transform
    vec = np.log1p(vec)

    return vec
```

```python
import os, glob, numpy as np, re
from pyteomics import mzml
from lxml.etree import XMLSyntaxError

MZML_DIR = r"C:\Users\Sachin Sharma\Music\raw_80datasets"
OUT_DIR  = os.path.join(MZML_DIR, "processed_bins")

os.makedirs(OUT_DIR, exist_ok=True)

# robust file detection (handles mzML and mzml safely)
files = sorted({
    os.path.normcase(p)
    for p in glob.glob(os.path.join(MZML_DIR, "*.mzML")) +
             glob.glob(os.path.join(MZML_DIR, "*.mzml"))
})

print("Total mzML detected:", len(files))

MZ_MIN, MZ_MAX, DELTA_M = 300.0, 1800.0, 1.0
N_BINS = int((MZ_MAX - MZ_MIN) / DELTA_M)

def label_from_name(name):
    m = re.search(r'(\d+)([TN])', name, flags=re.IGNORECASE)
    if not m:
        raise ValueError(f"No label found in {name}")
    return 1 if m.group(2).upper() == "T" else 0

def bin_file(path):
    vec = np.zeros(N_BINS, dtype=np.float64)
    with mzml.MzML(path) as reader:
        for spec in reader:
            if spec.get("ms level") != 1:
                continue
            mzs = spec.get("m/z array")
            ints = spec.get("intensity array")
            if mzs is None:
                continue
            mask = (mzs >= MZ_MIN) & (mzs < MZ_MAX)
            idx = ((mzs[mask] - MZ_MIN) / DELTA_M).astype(int)
            np.add.at(vec, idx, ints[mask])
    if vec.sum() > 0:
        vec /= vec.sum()
    return np.log1p(vec).astype(np.float32)

saved = 0
skipped = 0

for i, f in enumerate(files, 1):
    base = os.path.basename(f)
    vec_path = os.path.join(OUT_DIR, base + ".npy")
    lab_path = os.path.join(OUT_DIR, base + ".label.txt")

    if os.path.exists(vec_path) and os.path.exists(lab_path):
        continue

    try:
        x = bin_file(f)
        y = label_from_name(base)
        np.save(vec_path, x)
        with open(lab_path, "w") as fp:
            fp.write(str(y))
        saved += 1
        print(f"[{i}/{len(files)}] processed:", base)

    except XMLSyntaxError:
        skipped += 1
        print(f"[{i}/{len(files)}] broken skipped:", base)

print("\nNew files processed:", saved)
print("Broken skipped:", skipped)
```

```python
import numpy as np
from sklearn.feature_selection import SelectKBest, f_classif

# Load data
X = np.load("data/X.npy")
y = np.load("data/y.npy")

n_samples, n_features = X.shape

print("Samples:", n_samples)
print("Features:", n_features)

# Parameters
repeats = 50
subsample_ratio = 0.8
k_features = 200

# Stability counter
counts = np.zeros(n_features)

# Stability selection loop
for r in range(repeats):

    # random subset of samples
    idx = np.random.choice(n_samples, int(subsample_ratio*n_samples), replace=False)

    X_sub = X[idx]
    y_sub = y[idx]

    # ANOVA feature selection
    selector = SelectKBest(score_func=f_classif, k=k_features)
    selector.fit(X_sub, y_sub)

    selected_bins = selector.get_support(indices=True)

    # increase counts
    counts[selected_bins] += 1

    print(f"Repeat {r+1}/{repeats} done")

# Convert to stability scores
stability_scores = counts / repeats

# Select stable bins
threshold = 0.6
stable_bins = np.where(stability_scores >= threshold)[0]

print("\nStable bins found:", len(stable_bins))

# Save results
np.save("data/stability_scores.npy", stability_scores)
np.save("data/stable_bins.npy", stable_bins)

print("Saved stability_scores.npy and stable_bins.npy")
```

```python
import numpy as np

base = r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins"
stable_bins_fine = np.load(base + r"\stable_bins.npy")  # length 153, indices in [0..1499]

MZ_MIN = 300.0

def mz_range_from_multi_feature(j: int):
    """
    Returns: (feature_type, mz_low, mz_high)
    """
    # Block 1: X_stable (153 features) -> mapped via stable_bins_fine
    if 0 <= j < 153:
        fine_bin = int(stable_bins_fine[j])  # bin index in 0..1499
        mz_low = MZ_MIN + fine_bin * 1.0
        mz_high = mz_low + 1.0
        return ("fine(Δm=1, stable)", mz_low, mz_high)

    # Block 2: bins5 (300 features)
    if 153 <= j < 153 + 300:
        k = j - 153
        mz_low = MZ_MIN + k * 5.0
        mz_high = mz_low + 5.0
        return ("medium(Δm=5)", mz_low, mz_high)

    # Block 3: bins10 (150 features)
    k = j - (153 + 300)
    mz_low = MZ_MIN + k * 10.0
    mz_high = mz_low + 10.0
    return ("coarse(Δm=10)", mz_low, mz_high)
```

```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold
from sklearn.inspection import permutation_importance

base = r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins"
X = np.load(base + r"\X_multi_stable.npy")
y = np.load(base + r"\y.npy")
stable_feats_multi = np.load(base + r"\stable_feats_multi.npy")  # indices into X_multi (603)

pipe = Pipeline([
    ("scale", StandardScaler()),
    ("svm", SVC(kernel="rbf", probability=True, random_state=42))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Fit on full data (for a single global importance view)
pipe.fit(X, y)

perm = permutation_importance(
    pipe, X, y,
    scoring="roc_auc",
    n_repeats=30,
    random_state=42
)

imp = perm.importances_mean  # length = 198 (features in X_multi_stable)
top = np.argsort(imp)[::-1][:20]

print("Top 20 permutation important features (on X_multi_stable):")
for r, local_j in enumerate(top, 1):
    global_feat = int(stable_feats_multi[local_j])  # map back to 0..602 in X_multi
    t, lo, hi = mz_range_from_multi_feature(global_feat)
    print(f"{r:>3} | local={local_j:>3} | imp={imp[local_j]:.4f} | global={global_feat:>3} | {t} | {lo:.1f}-{hi:.1f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# use the importance array from permutation importance
mz_positions = []
importance_vals = []

for local_j in top:
    global_feat = stable_feats_multi[local_j]
    t, lo, hi = mz_range_from_multi_feature(global_feat)
    mz = (lo + hi)/2
    mz_positions.append(mz)
    importance_vals.append(imp[local_j])

plt.figure(figsize=(10,4))
plt.scatter(mz_positions, importance_vals)

plt.xlabel("m/z")
plt.ylabel("Feature Importance")
plt.title("Important Spectral Regions for Tumor Classification")

plt.show()
```

```python
import numpy as np

X = np.load(r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins\X_multi_stable.npy")
y = np.load(r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins\y.npy")

print("Dataset shape:", X.shape)
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

import matplotlib.pyplot as plt
```

```python
models = {
    "Logistic Regression": Pipeline([
        ("scale", StandardScaler()),
        ("model", LogisticRegression(max_iter=5000))
    ]),
    
    "SVM (RBF)": Pipeline([
        ("scale", StandardScaler()),
        ("model", SVC(kernel="rbf", probability=True))
    ]),
    
    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        random_state=42
    )
}
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

plt.figure(figsize=(8,6))

for name, model in models.items():
    
    tprs = []
    aucs = []
    mean_fpr = np.linspace(0,1,100)

    for train, test in cv.split(X,y):
        
        model.fit(X[train], y[train])
        
        probs = model.predict_proba(X[test])[:,1]
        
        fpr, tpr, _ = roc_curve(y[test], probs)
        roc_auc = auc(fpr,tpr)
        
        aucs.append(roc_auc)
        tprs.append(np.interp(mean_fpr, fpr, tpr))
        tprs[-1][0] = 0.0

    mean_tpr = np.mean(tprs, axis=0)
    mean_auc = auc(mean_fpr, mean_tpr)

    plt.plot(mean_fpr, mean_tpr,
             label=f"{name} (AUC = {mean_auc:.2f})")

plt.plot([0,1],[0,1],'k--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

base = r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins"

# Load stable fine-bin matrix + labels
X = np.load(base + r"\X_stable.npy")          # (68,153)
y = np.load(base + r"\y.npy")                 # (68,)

# We need stable bin indices to map columns -> actual m/z range
stable_bins = np.load(base + r"\stable_bins.npy")  # length 153, values in [0..1499]

MZ_MIN = 300.0
DELTA_M = 1.0

# Convert stable bin indices to m/z centers for plotting
mz_centers = MZ_MIN + stable_bins * DELTA_M + (DELTA_M / 2)

# ---- 1) Prepare a Z-scored matrix for heatmap ----
# Z-score per feature (column) so heatmap highlights relative differences
col_mean = X.mean(axis=0, keepdims=True)
col_std = X.std(axis=0, keepdims=True) + 1e-12
Xz = (X - col_mean) / col_std

# Sort samples by label (0=Normal first, then 1=Tumor)
order = np.argsort(y)
Xz_sorted = Xz[order]
y_sorted = y[order]

# ---- 2) Compute class mean spectra (original scale) ----
tum = (y == 1)
nor = (y == 0)

mean_tum = X[tum].mean(axis=0)
mean_nor = X[nor].mean(axis=0)
diff = mean_tum - mean_nor  # Tumor - Normal

# ---- 3) Plot: Heatmap + Mean curves + Difference ----
plt.figure(figsize=(12, 10))

# (A) Heatmap
ax1 = plt.subplot(3, 1, 1)
im = ax1.imshow(Xz_sorted, aspect='auto', interpolation='nearest')
ax1.set_title("Spectral Fingerprint Heatmap (Z-scored) on Stable Fine Bins")
ax1.set_ylabel("Samples (Normal → Tumor)")
# Put a boundary line between Normal and Tumor in sorted order
boundary = np.sum(y_sorted == 0)
ax1.axhline(boundary - 0.5)

cbar = plt.colorbar(im, ax=ax1, fraction=0.02, pad=0.02)
cbar.set_label("Z-score")

# X-axis ticks as m/z (few ticks so it stays readable)
tick_count = 10
tick_pos = np.linspace(0, X.shape[1]-1, tick_count).astype(int)
ax1.set_xticks(tick_pos)
ax1.set_xticklabels([f"{mz_centers[i]:.0f}" for i in tick_pos])

# (B) Mean spectra overlay
ax2 = plt.subplot(3, 1, 2)
ax2.plot(mz_centers, mean_nor, label="Normal mean")
ax2.plot(mz_centers, mean_tum, label="Tumor mean")
ax2.set_title("Class Mean Spectra (Stable Fine Bins)")
ax2.set_ylabel("Mean intensity (binned)")
ax2.legend()

# (C) Difference curve
ax3 = plt.subplot(3, 1, 3)
ax3.plot(mz_centers, diff)
ax3.axhline(0)
ax3.set_title("Difference Spectrum (Tumor − Normal)")
ax3.set_xlabel("m/z")
ax3.set_ylabel("Mean difference")

plt.tight_layout()

# Save figure for paper
outpath = base + r"\figure_heatmap_fingerprint.png"
plt.savefig(outpath, dpi=300)
plt.show()

print("Saved figure:", outpath)
```

```python
import numpy as np
from sklearn.model_selection import permutation_test_score, StratifiedKFold
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

X = np.load(r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins\X_multi_stable.npy")
y = np.load(r"C:\Users\Sachin Sharma\Music\raw_80datasets\processed_bins\y.npy")

model = Pipeline([
    ("scale", StandardScaler()),
    ("svm", SVC(kernel="rbf", probability=True))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

score, perm_scores, pvalue = permutation_test_score(
    model, X, y,
    scoring="roc_auc",
    cv=cv,
    n_permutations=200,
    random_state=42,
    n_jobs=-1
)

print("Observed ROC:", score)
print("Permutation mean ROC:", perm_scores.mean())
print("p-value:", pvalue)
```
